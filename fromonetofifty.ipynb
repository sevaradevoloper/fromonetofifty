```python
# DARS 8-14 Ehtimollik va Statistika Poydevori

# EHTIMOLLIK



# tajriba: Zar otamiz
# natija: 1 dan 6 gacha son tushadi
# Sample space(omega) - barcha ehtimolik tushushi mumkin bolgan natijalar toplami

# omega={soglom,kasal}
# omega={gerb,raqam}
# omega = {1,2,3,4,5,6}

# --- KOD: Zar misoli uchun Omegani to'plam sifatida yaratamiz ---
omega_zar = {1, 2, 3, 4, 5, 6}


# P(A) bu ehtimolik hodisa,yani A ning sodir bolish ehtimoliki
# A (tanga tashlaganimizada gerb tushish ehtimolliki qancha? P(A)=?)
# 3 marta tanga tashladik omega = {gerb,raqam,gerb}
# P(A) = A hodisa(omega)soni / barcha ehtimoli soniga(len(omega))
# P(A) = 2/3
# print(P)

# --- KOD: Siz yozgan 3 marta tanga tashlash misolini kodga o'tkazamiz ---
A_hodisa = 2
barcha_ehtimollar_soni = 3

P_A = A_hodisa / barcha_ehtimollar_soni
print("P(A) ehtimollik natijasi:", P_A)


# Union (Birlashma)( A ∪ B) OR qoidasi A yoki B hodisa bolsa
# Intersection (Kesishma)(A ∩ B) AND qoidasi A va B hodisa bolsa
# Addition Rule (Qo'shish qoidasi) P(A ∪ B) = P(A)+ P(B) - P(A ∩ B) bu degani A va B shartlarini qoshganimizada,hodisa 2x bolib ketmasligi uchun P(A ∩ B) qilib qoyamiz:

# --- KOD: Addition Rule qoidasini amalda tekshirish uchun zar misolida kod yozamiz ---
A = {2, 4, 6}
B = {4, 5, 6}

kesishma_AND = A & B   # Python-da (A ∩ B) -> {4, 6}
birlashma_OR = A | B    # Python-da (A ∪ B) -> {2, 4, 5, 6}

P_A_single = len(A) / len(omega_zar)
P_B_single = len(B) / len(omega_zar)
P_Intersection = len(kesishma_AND) / len(omega_zar)

P_Union_formula = P_A_single + P_B_single - P_Intersection
print("Addition Rule bo'yicha ehtimollik:", P_Union_formula)


# Hodisa independtent va depentdend boladi


# Conditional Probability (Shartli Ehtimollik) P(A|B)
# Bu degani: B hodisa sodir bo'lgani ma'lum bo'lganda, A hodisaning sodir bo'lish ehtimoli.
# Formula: P(A|B) = P(A ∩ B) / P(B)
# Masalan: Zarni tashlaganimizda 3 dan katta son tushgani aniq (B hodisa).
# Shu sonning JUFT bo'lish (A hodisa) ehtimolligi qancha? P(A|B) = ?

# --- KOD: Shartli ehtimollikni formula orqali hisoblaymiz ---
# Yuqoridagi P_Intersection ya'ni P(A ∩ B) va P_B_single ya'ni P(B) lardan foydalanamiz:
P_A_given_B = P_Intersection / P_B_single

print("Shartli ehtimollik P(A|B) natijasi:", P_A_given_B)
# Natija: 0.666... (ya'ni 3 dan katta {4, 5, 6} sonlar ichida juftlari {4, 6} -> 2/3 qismini tashkil qiladi)


# Bayes teoremasi yangi shart qoshilganda ehtimollik ozgarishni korsatadi:
# yangi ehtimollik: eski ehtimolik * yangi faktlar togriligi / jami faktlar ehtimoli

# =========================== =========================== ===========================
# =========================== =========================== ===========================

# Tasodifiy Miqdor (Random Variable) X bu biza otqazgan tajribalani songa otqazish funksiyasidir
# Discrete Random Variables (Diskret tasodifiy miqdorlar) -> sanaladigan (kopincha butun sonlar: {2,3,4})
# Continuous Random Variables (Uzluksiz tasodifiy miqdorlar) -> olchanadigan (masalan:odami boyi 1.75,vazni 55.7kg shunaqa)


#  PMF, PDF va CDF (Ehtimollik Funksiyalari)

#  PMF (Probability Mass Function - Ehtimollik Massasi Funksiyasi) PMF faqat va faqat Diskret tasodifiy miqdorlar uchun ishlatiladi.
# formulasi: f(x) = P(X = x) tasodifiy miqdor X ning aniq bir x qiymatiga teng bolish ehtimoliki qancha ? degan savolga javob beradi
# Zar tosh otamiz 4 soni tushish ehtimoliki P(X = 4) = 1/6
# barcha ehtmollikla yegindisi har doim 1 ga teng boladi yani 1/6 ni 6 martadan hisoblab qoshibchiqsa = 1 boladi. bu degani shu 1dan 6 gacha bolgan sonladan bittasi aniq tushadi!

#  PDF (Probability Density Function - Ehtimollik Zichligi Funksiyasi)
# PDF faqat va faqat Uzluksiz (Continuous) tasodifiy miqdorlar uchun ishlatiladi.
# Uzluksiz miqdorlar - bular cheksiz ko'p mayda qiymatlarni qabul qila oladigan o'zgaruvchilar (Masalan: inson bo'yi, harorat, vaqt, vazn).
# Formulasi: f(x) = dF(x)/dx (matematik integrallash orqali hisoblanadi)
# DIQQAT: PDF uzluksiz miqdorning ANIQ BIR nuqtadagi ehtimolligini bermaydi!
# Uzluksiz miqdorda aniq bir qiymat kelish ehtimoli har doim NOLGA teng: P(X = 170) = 0.
# Chunki inson bo'yi roppa-rosa 170.00000000... sm bo'lishi imkonsiz (mikronigacha hisoblasak).
# Savol: Unda PDF nima degan savolga javob beradi?
# Javob: U ehtimollikning qaysi ORALIQDA (intervalda) zichroq to'planganini ko'rsatadi.
# Masalan: Inson bo'yi 170 sm dan 180 sm gacha bo'lish ehtimolligi grafik ostidagi yuza (integral) orqali hisoblanadi.
#  Barcha grafik ostidagi umumiy yuza (minus cheksizlikdan plyus cheksizlikgacha integral)
# har doim 1 ga (100%) teng bo'ladi. Bu - umumiy dunyodagi hamma odamlar qaysidir bo'yga ega ekanligini anglatadi.


#  CDF (Cumulative Distribution Function - Jamlanib boruvchi taqsimot funksiyasi)
# CDF ham Diskret (PMF), ham Uzluksiz (PDF) tasodifiy miqdorlar uchun birdek ishlatiladi.
# "Cumulative" so'zi lotinchadan "jamlanib boruvchi", "to'planib boruvchi" degan ma'noni anglatadi.
#  Formulasi: F(x) = P(X <= x)
# Tasodifiy miqdor X ning ma'lum bir x qiymatidan KICHIK yoki TENG bo'lish ehtimoli qancha? degan savolga javob beradi.
# Ya'ni, u chap tomondan boshlab o'sha nuqtagacha bo'lgan barcha ehtimolliklarni yig'ib (jamlab) keladi.
#  Zar toshi (Diskret) misolida:
# P(X <= 3) -> Tosh otganda 3 yoki undan kichik son tushish ehtimoli.
# Bu: P(X=1) + P(X=2) + P(X=3) = 1/6 + 1/6 + 1/6 = 3/6 = 0.5 (50%).
#  Inson bo'yi (Uzluksiz) misolida:
# P(X <= 170) -> Tasodifiy insonning bo'yi 170 sm yoki undan past bo'lish ehtimoli.
# Bu PDF grafikining eng chapidan boshlab to 170 nuqtasigacha bo'lgan yuzani bildiradi.
#  CDFning 3 ta eng qat'iy qoidasi:
# 1. Grafigi har doim 0 dan boshlanadi va 1 da tugaydi (chunki ehtimollik 100% dan oshmaydi).
# 2. Hech qachon kamaymaydi (doim o'suvchi yoki tekis chiziq bo'ladi, chunki ehtimolliklar faqat qo'shilib boradi).
# 3. Ikki oraliq dagi ehtimollikni topish uchun CDF qiymatlari bir-biridan ayiriladi: P(170 <= X <= 180) = F(180) - F(170).

import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# 1. Zar toshining barcha ehtimoliy natijalarini belgilaymiz (1 dan 6 gacha diskret sonlar)
natijalar = np.array([1, 2, 3, 4, 5, 6])

# 2. scipy.stats orqali tekis diskret taqsimot (Discrete Uniform Distribution) yaratamiz
# Zar toshida har bir raqamning chiqish ehtimoli teng (1/6) bo'lgani uchun shu taqsimot tanlanadi
zar_taqsimoti = stats.randint(1, 7)  # 1 dan 7 gacha (7 kirmaydi, ya'ni 1,2,3,4,5,6)

# 3. Har bir natija uchun PMF, ya'ni P(X = x) aniq ehtimollik qiymatlarini hisoblaymiz
pmf_qiymatlari = zar_taqsimoti.pmf(natijalar)

# 4. Konsolga natijalarni chiroyli qilib chiqaramiz
print("--- ZAR TOSHI UCHUN PMF QIYMATLARI ---")
for x, p in zip(natijalar, pmf_qiymatlari):
    print(f"P(X = {x}) = {p:.4f}  (ya'ni 1/6)")

# Barcha ehtimolliklar yig'indisini tekshiramiz
yigindi = np.sum(pmf_qiymatlari)
print(f"\nBarcha ehtimolliklar yig'indisi: {yigindi:.1f}")

# 5. Grafik (Visualizatsiya) qismini chizamiz
plt.figure(figsize=(8, 5))

# Diskret taqsimot bo'lgani uchun ustunli grafik (bar) chizamiz
plt.bar(
    natijalar,
    pmf_qiymatlari,
    color="skyblue",
    edgecolor="black",
    alpha=0.8,
    width=0.4,
)

# Grafik sozlamalari
plt.title(
    "Zar toshini otish uchun PMF (Ehtimollik Massasi Funksiyasi)", fontsize=14
)
plt.xlabel("Tushgan raqam (X)", fontsize=12)
plt.ylabel("Ehtimollik P(X = x)", fontsize=12)
plt.xticks(natijalar)  # X o'qida faqat 1,2,3,4,5,6 sonlarini ko'rsatish
plt.ylim(0, 0.25)  # Y o'qi chegarasi (0.166 aniq ko'rinishi uchun)
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Har bir ustun tepasiga ehtimollik qiymatini yozib qo'yamiz
for x, p in zip(natijalar, pmf_qiymatlari):
    plt.text(x, p + 0.005, f"{p:.3f}", ha="center", fontsize=10, weight="bold")

# Grafikni ekranga chiqarish
plt.show()
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# 1. Uzluksiz Normal taqsimot yaratamiz (O'rtacha bo'y = 170, standart og'ish = 10)
boy_taqsimoti = stats.norm(loc=170, scale=10)

# 2. Grafik uchun X o'qidagi qiymatlarni (bo'y uzunliklarini) yaratamiz (140 dan 200 gacha)
x = np.linspace(140, 200, 500)
pdf_qiymatlari = boy_taqsimoti.pdf(x)  # PDF funksiyasini chaqiramiz

# 3. Aniq oraliqdagi ehtimollikni hisoblaymiz (Masalan: bo'yi 170 dan 180 sm gacha bo'lganlar)
# Buning uchun integral hisoblanadi (kodda cdf orqali oson hisoblanadi)
ehtimollik_170_180 = boy_taqsimoti.cdf(180) - boy_taqsimoti.cdf(170)
print(f"Bo'y 170 va 180 sm oralig'ida bo'lish ehtimolligi: {ehtimollik_170_180:.4f} (ya'ni {ehtimollik_170_180*100:.2f}%)")

# 4. Grafik chizamiz
plt.figure(figsize=(9, 5))
plt.plot(x, pdf_qiymatlari, color="blue", linewidth=2, label="PDF Chizig'i")

# 170 dan 180 gacha bo'lgan joyni bo'yaymiz (Zichlik/Yuza ehtimollikni beradi)
x_shart = np.linspace(170, 180, 100)
plt.fill_between(
    x_shart,
    boy_taqsimoti.pdf(x_shart),
    color="orange",
    alpha=0.4,
    label="P(170 <= X <= 180)",
)

plt.title(
    "Inson bo'yi misolida PDF (Ehtimollik Zichligi Funksiyasi)", fontsize=14
)
plt.xlabel("Bo'y (sm)", fontsize=12)
plt.ylabel("Zichlik (Density)", fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()


import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# 1. Uzluksiz Normal taqsimot (O'rtacha bo'y = 170, standart og'ish = 10)
boy_taqsimoti = stats.norm(loc=170, scale=10)

# 2. X o'qi qiymatlari va ular uchun CDF, ya'ni F(x) qiymatlarini hisoblaymiz
x = np.linspace(140, 200, 500)
cdf_qiymatlari = boy_taqsimoti.cdf(x)  # CDF funksiyasini chaqiramiz

# 3. Aniq nuqtadagi CDF qiymatini tekshiramiz
# Bo'yi 170 sm yoki undan past bo'lgan odamlar necha foiz?
p_past_170 = boy_taqsimoti.cdf(170)
print(f"Bo'yi 170 sm dan past bo'lish ehtimoli: {p_past_170:.4f} (ya'ni {p_past_170*100:.1f}%)")

# 4. Grafik chizamiz
plt.figure(figsize=(9, 5))
plt.plot(x, cdf_qiymatlari, color="green", linewidth=2.5, label="CDF Chizig'i F(x)")

# Grafikka o'rtacha qiymat nuqtasini belgilab qo'yamiz (O'rtacha nuqtada CDF doim 0.5 bo'ladi)
plt.scatter(170, 0.5, color="red", s=100, zorder=5)
plt.axhline(y=0.5, color="red", linestyle="--", alpha=0.5)
plt.axvline(x=170, color="red", linestyle="--", alpha=0.5)
plt.text(171, 0.45, "F(170) = 0.5 (50%)", color="red", weight="bold")

# Grafik sozlamalari
plt.title("Inson bo'yi misolida CDF Grafigi (O'suvchi ehtimollik)", fontsize=14)
plt.xlabel("Bo'y (sm)", fontsize=12)
plt.ylabel("Jamlangan Ehtimollik P(X <= x)", fontsize=12)
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()



# =========================== =========================== ===========================
# =========================== =========================== ===========================

# MALUMOTLARNI TAQSIMOTI

# Asosiy Uch asosiy taqsimot Normal,Binomial,Poisson

from scipy import stats

#  Uch Asosiy Taqsimot (Normal, Binomial, Poisson)

#  1. Normal Taqsimot (Normal Distribution)
# Turi: Uzluksiz (Continuous) taqsimot.
# Mohiyati: Tabiatdagi eng universal taqsimot (Qo'ng'iroqsimon grafik). Agar biror ko'rsatkichga
# juda ko'p mayda, tasodifiy va mustaqil omillar ta'sir qilsa, ma'lumotlar NORMAL taqsimlanadi.
# Parametrlari: loc (o'rtacha qiymat, mu) va scale (standart og'ish, sigma).
# •               O'rtacha qiymat ( 𝜇(mu)) va Standart og'ish ( σ(sigma)) kerak.
# Real misol: Insonlarning bo'yi, vazni, imtihon ballari, zavod mahsulotlarining o'lchami.

#  2. Binomial Taqsimot (Binomial Distribution)
# Turi: Diskret (Discrete) taqsimot.
# Mohiyati: Faqat IKKI XIL natijasi bor (Ha/Yo'q, Muvaffaqiyat/Mag'lubiyat) va bir-biriga
# bog'liq bo'lmagan chekli sondagi (n) sinovlar o'tkazilganda yutuqlar sonini hisoblaydi.
# Parametrlari: n (sinovlar soni) va p (bitta sinovdagi muvaffaqiyat ehtimoli).
# Real misol: Tangani 20 marta tashlaganda necha marta "Gerb" tushishi, 100 ta mijozdan nechtasi mahsulot sotib olishi.

#  3. Poisson Taqsimoti (Poisson Distribution)
# Turi: Diskret (Discrete) taqsimot.
# Mohiyati: Aniq bir VAQT yoki MAKON oralig'ida sodir bo'ladigan, lekin kamdan-kam ro'y beradigan
# tasodifiy voqealar sonini o'lchaydi. Voqealar doimiy o'rtacha tezlik (lambda) bilan sodir bo'ladi.
# Parametrlari: mu yoki lambda (ma'lum bir vaqt ichidagi o'rtacha voqealar soni).
# Real misol: 1 soat ichida do'konga kirgan xaridorlar soni, kitob sahifasidagi imlo xatolari soni.

import scipy.stats as stats

print("--- 1. NORMAL TAQSIMOT (Uzluksiz) ---")
# O'rtacha bo'y 170 sm, standart og'ish 10 sm bo'lgan aholi
normal_model = stats.norm(loc=170, scale=10)
print(f"Kutilgan o'rtacha bo'y E[X]: {normal_model.mean()} sm")
# Tasodifiy odamning bo'yi 180 sm dan baland bo'lish ehtimoli (PDF ustida sf funksiyasi)
print(f"Bo'y > 180 sm bo'lish ehtimoli: {normal_model.sf(180):.4f}\n")


print("--- 2. BINOMIAL TAQSIMOT (Diskret) ---")
# Tangani n=20 marta tashlaymiz, har safar 'Gerb' tushish ehtimoli p=0.5
binom_model = stats.binom(n=20, p=0.5)
print(f"Kutilgan yutuqlar soni E[X] (n * p): {binom_model.mean()}")
# ROPPA-ROSA 10 marta 'Gerb' tushish ehtimoli (Diskret bo'lgani uchun PMF ishlatiladi)
print(f"Roppa-rosa 10 marta chiqish ehtimoli P(X=10): {binom_model.pmf(10):.4f}\n")


print("--- 3. POISSON TAQSIMOTI (Diskret) ---")
# Do'konga 1 soatda o'rtacha lambda=4 ta xaridor kiradi
poisson_model = stats.poisson(mu=4)
print(f"1 soatda kutilgan o'rtacha xaridorlar E[X]: {poisson_model.mean()}")
# Kelgusi 1 soatda do'konga ROPPA-ROSA 6 ta xaridor kirish ehtimoli (PMF)
print(f"Roppa-rosa 6 ta xaridor kirish ehtimoli P(X=6): {poisson_model.pmf(6):.4f}")

#  BU TAQSIMOTLAR MACHINE LEARNINGDA QAYERDA ISHLATILADI?

#  1. Normal Taqsimot (Linear Models & Deep Learning)
#  Xatoliklarni tahlil qilish (Residuals): Chiziqli regressiya (Linear Regression)
# modelini qurgandan keyin, model yo'l qo'ygan xatoliklar (residuals) NORMAL TAQSIMLANGAN
# bo'lishi shart. Agar unday bo'lmasa, modelda muammo bor deb hisoblanadi.
#  Ma'lumotlarni normallashtirish (StandardScaler): Ko'plab algoritmlar (KNN, SVM,
# Neural Networks) ma'lumotlar normal taqsimlanganda eng yaxshi natijani beradi. Biz
# ularni mu=0 va sigma=1 holatiga keltirish uchun Standardizatsiya qilamiz.
#  Vae (Variational Autoencoders): Tasvirlarni generatsiya qiluvchi ilg'or neyron
# tarmoqlari yashirin joyni (latent space) normal taqsimot orqali modellashtiradi.

#  2. Binomial Taqsimot (Binary Classification & Metrics)
#  Logistik Regressiya (Logistic Regression): Ikki toifali klassifikatsiya (masalan,
# "Bemor / Sog'lom", "Spam / Spam emas") modellarining asosi aynan Binomial taqsimotga
# tayanadi. Model chiqargan ehtimollik (0.85) Binomial tajribaning 'p' parametridir.
#  A/B Testing: Sayt dizaynining qaysi biri yaxshiligini tekshirishda ML muhandislari
# konversiya (konversiya bo'ldi / bo'lmadi) natijalarini solishtirish uchun Binomial
# testlardan foydalanishadi.

#  3. Poisson Taqsimoti (Count Data & Forecasting)
#  Sanoq ma'lumotlarini bashorat qilish (Poisson Regression): Agar siz qurayotgan model
# "Ertaga do'konga nechta odam kiradi?", "Toshkentda ertaga nechta yo'l transport hodisasi
# sodir bo'ladi?" yoki "Saytga bir daqiqada nechta so'rov keladi?" degan SAVOLLARGA javob
# berishi kerak bo'lsa (ya'ni natija manfiy bo'lmagan butun sonlar bo'lsa), Chiziqli regressiya
# emas, balki Poisson Regressiya modeli ishlatiladi.


# =========================== =========================== ===========================
# =========================== =========================== ===========================

#  O'rtacha qiymat (Mean), Variatsiya (Variance) va Standart og'ish (Standard Deviation)

#  1. O'rtacha qiymat (Mean - \mu)
# Mohiyati: Ma'lumotlar to'plamining "markaziy nuqtasi" yoki kutilgan qiymati.
# Qanday hisoblanadi: Barcha qiymatlar yig'indisini ularning umumiy soniga bo'lamiz.
# MLda o'rni: Ma'lumotlardagi bo'shliqlarni (Missing Values) to'ldirishda (Imputation)
# yoki ma'lumotlarni markazlashtirishda (Centering) keng qo'llaniladi.

#  2. Variatsiya / Dispersiya (Variance - \sigma^2)
# Mohiyati: Ma'lumotlarning o'rtacha qiymatdan qanchalik uzoqlashganini (tarqoqligini) o'lchaydi.
# Qanday hisoblanadi: Har bir qiymatdan o'rtacha qiymat ayirilib, kvadrati olinadi va ularning o'rtachasi hisoblanadi.
# Muammosi: Qiymatlar kvadratga oshirilgani uchun, uning o'lchov birligi ham kvadrat bo'lib ketadi
# (Masalan, agar ma'lumot so'mda bo'lsa, Variatsiya "so'm kvadrat" bo'lib qoladi. Buni tasavvur qilish qiyin).
# MLda o'rni: PCA (Asosiy komponentlar tahlili) algoritmida eng ko'p ma'lumot saqlab qolgan
# o'qni topish uchun aynan Variatsiya maksimal darajaga ko'tariladi.

#  3. Standart og'ish (Standard Deviation - \sigma)
# Mohiyati: Variatsiyaning eng yaqin do'sti, lekin tushunishga ancha oson shakli.
# Qanday hisoblanadi: Shunchaki Variatsiyadan kvadrat ildiz (sqrt) olinadi.
# Natijada o'lchov birligi yana o'z holiga qaytadi (Masalan, "so'm").
# Qoida: Standart og'ish qanchalik KATTA bo'lsa, ma'lumotlar to'plami shunchalik tarqoq (sochilgan) bo'ladi.
# Qanchalik KICHIK bo'lsa, ma'lumotlar shunchalik o'rtacha qiymat atrofida zich joylashgan bo'ladi.
import numpy as np
import scipy.stats as stats

# 1. Sun'iy ma'lumotlar to'plami (ML talabalarining yoshi)
talabalar_yoshi = np.array([20, 22, 21, 24, 19, 30, 25, 23, 21, 28])

# 2. O'rtacha qiymatni hisoblaymiz (Mean)
mean_yosh = np.mean(talabalar_yoshi)
print(f"Talabalarning o'rtacha yoshi (Mean): {mean_yosh} yosh")

# 3. Variatsiyani hisoblaymiz (Variance)
# MLda ko'pincha tanlanma (sample) bilan ishlaganimiz uchun ddof=1 (Bessel's correction) beriladi
var_yosh = np.var(talabalar_yoshi, ddof=1)
print(f"Yoshlarning variatsiyasi (Variance): {var_yosh:.4f} (yosh kvadrat)")

# 4. Standart og'ishni hisoblaymiz (Standard Deviation)
std_yosh = np.std(talabalar_yoshi, ddof=1)
print(f"Standart og'ish (Standard Deviation): {std_yosh:.4f} yosh")

# Mantiqiy xulosa:
print(f"\nXulosa: Guruhdagi talabalarning o'rtacha yoshi {mean_yosh} da va ularning aksariyati "
      f"bu ko'rsatkichdan \u00b1{std_yosh:.1f} yosh atrofida farq qiladi.")



# =========================== =========================== ===========================
# =========================== =========================== ===========================
#  Population Variance vs Sample Variance (Bosh to'plam va Tanlanma dispersiyasi)

#  1. Population Variance (\sigma^2)
# Mohiyati: Dunyodagi/Olamdagi BARCHA ob'ektlar (Population) hisobga olingandagi dispersiya.
# Formula maxraji: N (Jami elementlar soni).
# Real misol: O'zbekistondagi barcha 37+ million aholining yoshlaridagi tarqoqlik.

#  2. Sample Variance (s^2)
# Mohiyati: Bosh to'plam ichidan ajratib olingan KICHIK GURUH (Tanlanma / Sample) dispersiyasi.
# Formula maxraji: n - 1 (Bessel tuzatishi - Bessel's Correction).
# Real misol: 37 million aholidan tasodifiy tanlab olingan 1000 ta odamning yoshi.

#  NEGA N EMAS, n - 1 GA BO'LAMIZ?
# Biz kichik guruhni olganimizda, odatda chekka (ekstremal) qiymatlarni hisobdan chiqarib yuboramiz.
# Agar shunchaki n ga bo'lsak, biz topgan dispersiya haqiqiy (Population) dispersiyadan KICHIK bo'lib qoladi.
# Matematiklar maxrajni n - 1 qilib, maxrajni kichraytirishadi. Maxraj kichraygach, yakuniy natija (s^2)
# biroz KATTALASHADI va haqiqiy dunyodagi dispersiyani to'g'ri ( xolis / unbiased) bashorat qiladi.

#  MLda o'rni: Biz deyarli har doim Sample Variance (n-1) bilan ishlaymiz, chunki qo'limizda hech
# qachon dunyodagi hamma ma'lumotlar (Population) bo'lmaydi, faqat ma'lum bir dataset (Sample) bo'ladi.


import numpy as np
import scipy.stats as stats

# 1. Ma'lumotlar to'plami (Masalan, 5 ta uyning narxi, million so'mda)
uy_narxlari = np.array([300, 450, 280, 600, 370])

print("--- 1. DISPERSIYA FARQLARI ---")
# Population Variance (ddof=0, standart numpy holati)
pop_var = np.var(uy_narxlari, ddof=0)
print(f"Population Variance (\u03c3\u00b2) [n ga bo'lingan]: {pop_var:.2f}")

# Sample Variance (ddof=1, MLda ishlatiladigan xolis variant)
sample_var = np.var(uy_narxlari, ddof=1)
print(f"Sample Variance (s\u00b2) [n-1 ga bo'lingan]: {sample_var:.2f}")


print("\n--- 2. Z-SCORE HISOBLASH ---")
# numpy yoki scipy orqali har bir uy narxining Z-score qiymatini topamiz
z_scores = stats.zscore(uy_narxlari)

for narx, z in zip(uy_narxlari, z_scores):
    print(f"Uy narxi: {narx} mln | Z-score: {z:+.4f}")

print("\n💡 Izoh: 600 mlnlik uyning Z-score ko'rsatkichi eng yuqori (+1.71), "
      "chunki u o'rtacha narxdan ancha qimmat.")


#  Z-Score (Standartlashtirilgan baho)

# Mohiyati: Biror ma'lumot (nuqta) o'z guruhining o'rtacha qiymatidan (Mean) necha qadam
# (Standart og'ish - Standard Deviation) uzoqlikda joylashganini ko'rsatuvchi son.

#  Formulasi: Z = (X - \mu) / \sigma
# Bu yerda: X - qiymat, \mu - o'rtacha qiymat, \sigma - standart og'ish.

#  Z-score qiymatlarining ma'nosi:
# Z = 0 -> Qiymat roppa-rosa o'rtacha qiymatga teng.
# Z = +1.5 -> Qiymat o'rtacha ko'rsatkichdan 1.5 ta standart og'ish yuqorida (o'rtachadan katta).
# Z = -2.0 -> Qiymat o'rtacha ko'rsatkichdan 2 ta standart og'ish pastda (o'rtachadan ancha kichik).

#  MLda Z-Score qayerda ishlatiladi?
# 1. Outliers (Anomaliyalarni) topish: Agar biror ma'lumotning Z-score qiymati +3 dan katta
# yoki -3 dan kichik bo'lsa, bu o'ta g'alati/kam uchraydigan ma'lumot (Anomaliya) hisoblanadi va datasetdan o'chiriladi.
# 2. StandardScaler (Normallashtirish): Ko'plab ML algoritmlari (KNN, SVM, K-Means) ishlashi uchun hamma
# ustunlarni Z-score formulasiga solib, o'rtacha qiymatini 0, standart og'ishini 1 holatiga keltirib olamiz.

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 1. Sun'iy ma'lumotlar yaratamiz (bir nechta anomal qiymatlar bilan)
np.random.seed(42)
data = np.random.normal(loc=20, scale=5, size=100)
data = np.append(data, [45, 42, -5, -2])  # Outliers (g'ayritabiiy qiymatlar)

# Z-score hisoblaymiz
z_scores = stats.zscore(data)
threshold = 2  # G'ayritabiiy nuqtalarni belgilash chegarasi (|Z| > 2)

# Normal va anomal indekslarni ajratamiz
outliers_idx = np.where(np.abs(z_scores) > threshold)
normal_idx = np.where(np.abs(z_scores) <= threshold)

# 2. Grafik oynasini tayyorlaymiz (1 qator, 3 ta grafik)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.set_theme(style="whitegrid")

# --- 1-GRAFIK: Scatter Plot (Nuqtali grafik) ---
axes[0].scatter(normal_idx[0], data[normal_idx], color='blue', alpha=0.6, label='Normal', s=40)
axes[0].scatter(outliers_idx[0], data[outliers_idx], color='red', marker='X', s=100, label=f'Outliers (|Z| > {threshold})')
axes[0].set_title("Ma'lumotlar va Outliers")
axes[0].set_xlabel("Indeks")
axes[0].set_ylabel("Qiymat")
axes[0].legend()

# --- 2-GRAFIK: Histogramma va KDE (Z-score Taqsimoti) ---
sns.histplot(z_scores, kde=True, ax=axes[1], color='purple')
# Chegara chiziqlarini chizamiz
axes[1].axvline(x=threshold, color='red', linestyle='--', label=f'Z = +{threshold}')
axes[1].axvline(x=-threshold, color='red', linestyle='--', label=f'Z = -{threshold}')
axes[1].set_title("Z-Score Taqsimoti (Bell Curve)")
axes[1].set_xlabel("Z-score qiymati")
axes[1].legend()

# --- 3-GRAFIK: Boxplot (Mo'ylovli quti) ---
sns.boxplot(y=data, ax=axes[2], color='lightblue', width=0.4)
# Outlier nuqtalarini grafik ustiga qizil rangda qo'yamiz
axes[2].scatter(np.zeros(len(outliers_idx[0])), data[outliers_idx], color='red', marker='X', s=100, zorder=5)
axes[2].set_title("Boxplot orqali ko'rinishi")
axes[2].set_ylabel("Qiymat")

# Grafiklar bir-biriga yopishib qolmasligi uchun joyni to'g'rilaymiz
plt.tight_layout()
plt.show()

# Pandas da: Aytaylik sizda df degan jadval bor va price ustuni bor.
# ● df['price'].mean() -> O'rtachani beradi.
# ● df['price'].median() -> Medianani beradi.
# ● df['price'].mode() -> Modani beradi.
# ● df['price'].quantile(0.25) -> Q1 ni beradi (25th percentile). Q3 uchun esa 0.75 yozasiz.
# ● df['price'].var() -> Variatsiyani beradi.
# ● df['price'].std() -> Standart og'ishni beradi.
# ● df['price'].describe() -> Yuqoridagilarning ko'pchiligini bitta zo'r jadval qilib chiqarib beradi.


# =========================== =========================== ===========================
# =========================== =========================== ===========================
#  Markaziy Limit Teoremasi (Central Limit Theorem - CLT)

#  Teoremaning Oltin Qoidasi:
# Qo'limizdagi ma'lumotlar aslida qanday taqsimlangan bo'lishidan qat'iy nazar (xoh tekis,
# xoh eksponentsial, xoh mutlaqo tartibsiz bo'lsin), agar biz u yerdan yetarlicha katta hajmdagi
# (odatda n >= 30) tasodifiy tanlanmalar (samples) olib, ularning o'rtacha qiymatlarini (Sample Means)
# hisoblasak va o'sha o'rtacha qiymatlarni grafikga chizsak — u HAR DOIM NORMAL TAQSIMOT (qo'ng'iroq simon) bo'ladi!

#  CLTning 3 ta asosiy mo'jizasi:
# 1. Tanlanmalar o'rtacha qiymatining o'rtachasi (Mean of Sample Means) haqiqiy Bosh to'plamning
# o'rtacha qiymatiga (\mu) teng bo'ladi.
# 2. Tanlanma hajmi (n) qanchalik katta bo'lsa, o'rtacha qiymatlar grafigi shunchalik zichlashadi
# va aniqlik ortadi (Standart xatolik - Standard Error kamayadi: \sigma / \sqrt{n}).
# 3. Bu teorema bizga real dunyodagi har qanday tartibsiz dataset ustida ham Normal taqsimot
# qonuniyatlarini va gipoteza testlarini (Z-test, T-test) qo'llash imkonini beradi.

#  MLda o'rni: Bootstrapping usulida (masalan, Random Forest algoritmida ma'lumotlarni
# ko'paytirishda) va modellarning barqarorligini tekshirishda (Cross-Validation) CLT qonuniyatlari yotadi.
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# 1. Mutlaqo tartibsiz (Tekis) taqsimlangan 100,000 ta ma'lumot yaratamiz
bosh_toplam = np.random.uniform(0, 100, 100000)

# 2. CLT simulyatsiyasi: 1000 marta, har safar 40 tadan tasodifiy ma'lumot olib, o'rtachasini hisoblaymiz
tanlanma_ortachalari = []
for _ in range(1000):
    tanlanma = np.random.choice(bosh_toplam, size=40)
    tanlanma_ortachalari.append(np.mean(tanlanma))

# 3. Grafiklarni chizish
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Birinchi grafik - Haqiqiy bosh to'plam (Tartibsiz, tekis g'isht shaklida)
ax1.hist(bosh_toplam, bins=30, color='gray', alpha=0.6, edgecolor='black')
ax1.set_title("Haqiqiy Bosh To'plam (Uniform)")

# Ikkinchi grafik - O'rtacha qiymatlar taqsimoti (Mo'jizaviy Normal qo'ng'iroq shakli!)
ax2.hist(tanlanma_ortachalari, bins=30, color='blue', alpha=0.6, edgecolor='black', density=True)
ax2.set_title("Tanlanmalar O'rtacha Qiymati (CLT natijasi)")

plt.tight_layout()
plt.show()

#  Kovariatsiya va Korrelyatsiya (Covariance & Correlation)

#  1. Kovariatsiya (Covariance - Cov(X, Y))
# Mohiyati: Ikki o'zgaruvchining birgalikda qay darajada o'zgarayotganini (yo'nalishini) ko'rsatadi.
#  Cov(X, Y) > 0 -> Ijobiy yo'nalish. X ortsa, Y ham ortadi (Masalan: Reklama budjeti va Sotuv).
#  Cov(X, Y) < 0 -> Salbiy yo'nalish. X ortsa, Y kamayadi (Masalan: Mashina yoshi va uning narxi).
#  Cov(X, Y) = 0 -> Ular o'rtasida hech qanday bog'liqlik yo'q.
# Katta kamchiligi: Uning chegarasi yo'q (Qiymati -cheksizlikdan +cheksizlikgacha bo'lishi mumkin).
# Shuning uchun bog'liqlik kuchi qanchalik baquvvat ekanini tushunib bo'lmaydi.

#  2. Korrelyatsiya (Pirson koeffitsiyenti - Pearson's r)
# Mohiyati: Kovariatsiyaning "mukammallashtirilgan", standartlashtirilgan shakli.
# U Kovariatsiyani har ikkala o'zgaruvchining standart og'ishiga bo'lish orqali olinadi.
# Chegarasi: Qat'iy -1 dan +1 gacha bo'lgan oraliqda bo'ladi!
#  r = +1 -> Mukammal to'g'ri chiziqli ijobiy bog'liqlik.
#  r = -1 -> Mukammal to'g'ri chiziqli salbiy bog'liqlik.
#  r = 0  -> Hech qanday chiziqli bog'liqlik yo'q.

# ==============================================================================
#  KORRELYATSIYANING 3 TA ASOSIY KOEFFITSIYENTI (PEARSON, SPEARMAN, KENDALL)
# ==============================================================================

#  1. Pirson Korrelyatsiyasi (Pearson Correlation Coefficient - r)
#  Qachon ishlatiladi: Faqat va faqat ikki o'zgaruvchi o'rtasida CHIZIQLI (Linear) bog'liqlik bo'lsa.
#  Mantiqiy sharti: Ma'lumotlar uzluksiz (continuous) va Normal taqsimlangan bo'lishi kerak.
#  Hayotiy misol: Avtomobil motorining hajmi va uning yoqilg'i sarfi (chiziqli ko'tariladi).
#  Kamchiligi: Outliers (anomaliyalar) ga juda sezgir. Agar bitta ma'lumot g'alati bo'lsa, chizig'i buziladi.

#  2. Spirman Korrelyatsiyasi (Spearman Rank Correlation - \rho)
#  Qachon ishlatiladi: Ma'lumotlar chiziqli bo'lmasa, lekin MONOTON bo'lsa (ya'ni X oshsa, Y ham doim oshadi,
# lekin to'g'ri chiziq bo'ylab emas, egri chiziq bo'ylab).
#  Mantiqiy sharti: U raqamlarning o'zi bilan emas, balki ularning O'RNI (Rank/Darajasi) bilan ishlaydi!
#  Hayotiy misol: O'qishga sarflangan vaqt va imtihondan olingan reyting o'rni (1-o'rin, 2-o'rin...).
#  Ustunligi: Anomaliyalarga (Outliers) judayam chidamli. Chunki u raqam kattaligiga emas, o'rniga qaraydi.

#  3. Kendall Tau Korrelyatsiyasi (Kendall's \tau)
#  Qachon ishlatiladi: Xuddi Spirman kabi ma'lumotlar tartiblangan (Ordinal) yoki monoton bo'lganda.
#  Mantiqiy sharti: U juftliklarni bir-biriga solishtiradi (Concordant va Discordant juftliklar).
# Ya'ni, "Agar X oshayotganda Y ham oshayotgan juftliklar soni ko'pmi yoki teskarisimi?" deb hisoblaydi.
#  Ustunligi: Agar datasetingiz judayam KICHIK bo'lsa, Kendall koeffitsiyenti Spirmandan ko'ra aniqroq
# va barqarorroq natija beradi.


#  ENG OLTI QOIDA: "Korrelyatsiya sabab-natija emas!" (Correlation is not causation)
# Ikki o'zgaruvchining birgalikda o'zgarishi (r=0.9) ulardan biri ikkinchisini keltirib
# chiqarganini anglatmasligi mumkin (Ular shunchaki uchinchi bir yashirin omil tufayli tebranayotgan bo'lishi mumkin).

#  MLda o'rni: Feature Selection (Ustunlarni saralash) jarayonida o'zaro juda yuqori korrelyatsiyaga
# ega bo'lgan (Multicollinearity) ortiqcha ustunlarni modeldan olib tashlash uchun ishlatiladi.
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Sun'iy ma'lumotlar yaratamiz
reklama_budjeti = np.array([10, 20, 30, 40, 50, 60, 70])
mahsulot_sotuvi = np.array([15, 25, 33, 45, 52, 68, 75])

# DataFrame yaratamiz
df = pd.DataFrame({'Reklama': reklama_budjeti, 'Sotuv': mahsulot_sotuvi})

# 2. Kovariatsiya matritsasini hisoblaymiz
kovariatsiya_matrix = np.cov(reklama_budjeti, mahsulot_sotuvi)
print(f"Kovariatsiya qiymati (Cov(X,Y)): {kovariatsiya_matrix[0, 1]:.2f}")

# 3. Korrelyatsiya matritsasini hisoblaymiz (Pandas orqali juda oson)
korrelyatsiya_matrix = df.corr(method='pearson')
print(f"Pirson Korrelyatsiya koeffitsiyenti (r): {korrelyatsiya_matrix.loc['Reklama', 'Sotuv']:.4f}")

# 4. Heatmap (Issiqlik xaritasi) orqali vizualizatsiya qilamiz
plt.figure(figsize=(5, 4))
sns.heatmap(korrelyatsiya_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Korrelyatsiya Heatmap grafigi")
plt.show()


# =========================== =========================== ===========================
# =========================== =========================== ===========================

# ==============================================================================
#   GIPOTEZALARNI TEKSHIRISH (HYPOTHESIS TESTING)
# ==============================================================================

# ------------------------------------------------------------------------------
#  1. GIPOTEZA NIMA? (Null vs Alternative Hypothesis)
# ------------------------------------------------------------------------------
# Muammo: Tasavvur qiling, siz saytdagi tugmaning rangini ko'kdan yashilga o'zgartirdingiz.
# Keyin qarasangiz, sotuvlar 2% ga oshibdi. Savol: Bu rostdan ham rang almashgani uchunmi
# (qonuniyat) yoki shunchaki o'sha kuni tasodifan odamlar ko'proq narsa sotib olishganmi (tasodif)?
# Statistika mana shu shubhani hal qilish uchun dunyoga kelgan.

# Biz har doim ikkita bir-biriga dushman gipoteza (fikr) bilan ish boshlaymiz:

#  A. Null Hypothesis (Nol Gipoteza - H0):
#  Mantiqiy ma'nosi: "Dunyoda hech qanday mo'jiza sodir bo'lgani yo'q. Hamma narsa eski hammom, eski tos."
#  Bu gipoteza har doim o'zgarish yo'qligini, farq nolga tengligini himoya qiladi.
#  Real misol: Tugma rangini yashil qilish sotuvga HECH QANDAY ta'sir qilmadi. O'sha 2% lik o'sish — shunchaki tasodif!
#  MLda o'rni: Siz yaratgan yangi qimmat model eski sodda modeldan umuman yaxshi emas.

#  B. Alternative Hypothesis (Muqobil Gipoteza - H1 yoki Ha):
#  Mantiqiy ma'nosi: "To'xtanglar, bu yerda qandaydir yangilik bor! Farq shunchaki tasodif emas!"
#  Bu gipoteza siz isbotlamoqchi bo'lgan haqiqatdir.
#  Real misol: Tugma rangining yashil bo'lishi odamlarga ko'proq yoqdi va sotuvni ROSTDAN HAM oshirdi!
#  MLda o'rni: Yangi model eski modelga qaraganda ancha aqlli va aniqroq bashorat qilyapti.


# ------------------------------------------------------------------------------
#  2. P-VALUE (Ehtimollik ko'rsatkichi) — "Tasodif o'lchagich"
# ------------------------------------------------------------------------------
# Muammo: Xo'sh, H0 to'g'rimi yoki H1? Buni qaysi tarozi o'lchaydi? Javob: P-Value.
#  Oddiy tilda: p-value — bu "Hamma narsa tasodif" (H0) degan gapning qanchalik haqiqatga yaqinligini ko'rsatuvchi foizdir.

# Tasavvur qiling, bir bola kelib: "Men tangani tashlasam doim Gerb tushadi, menda sehrli kuch bor!" (H1) dedi.
# Siz aytdingiz: "Yo'q, sen shunchaki omadli bolasan, hamma narsa tasodif" (H0).
# Bola tangani 5 marta tashladi va beshalasida ham GERB tushdi.

# Hisoblaymiz: Tangani 5 marta tashlaganda besh safar ham tasodifan Gerb tushish ehtimoli:
# (1/2) * (1/2) * (1/2) * (1/2) * (1/2) = 1/32 = 0.031 (ya'ni 3.1%).
# Mana shu 3.1% — bizning p-value qiymatimizdir!

# Endi qaror qabul qilamiz. Statistiklar o'rtasida "Alfa (\alpha) = 0.05 (5%)" degan qat'iy chegara bor:
#  Agar p-value < 0.05 bo'lsa (Bizning misolda 3.1% < 5%):
#  Bu degani: Natijaning tasodifan sodir bo'lish ehtimoli o'ta past (5% dan kichik). Demak, H0 o'yinni yutqazdi!
#  Biz H0 ni rad etamiz (Reject H0) va bolaning sehrli kuchi borligiga (H1) ishonishga majbur bo'lamiz.
#  Natija: "Statistik jihatdan ahamiyatli" (Statistically Significant).

#  Agar p-value >= 0.05 bo'lsa:
#  Bu degani: Natija shunchaki tasodifan ham sodir bo'lishi mumkin bo'lgan oddiy holat.
#  Biz H0 ni rad eta olmaymiz (Fail to reject H0). Bola shunchaki laqma, sehr-mehr yo'q!


# ------------------------------------------------------------------------------
#  3. ENGO'QP ISHLATILADIGAN STATISTIK TESTLAR (Qaysi birini qachon tanlaymiz?)
# ------------------------------------------------------------------------------
# Ma'lumotlaringiz qanday ko'rinishda ekanligiga qarab testlar 3 ga bo'linadi:

#  A. Z-Test vs T-Test (Ikki guruh o'rtacha qiymatini urishtirish)
# Qachon ishlatiladi: Sizda raqamli ma'lumotlar bor va siz faqatgina IKKI TA guruhni solishtirmoqchisiz.
#  Farqi:
#  Z-test: Agar siz butun dunyodagi (Population) standart og'ishni aniq bilsangiz va ma'lumotingiz ko'p bo'lsa (n >= 30).
#  T-test: Agar dunyo parametrlarini bilmasangiz (noma'lum bo'lsa) va faqat qo'lingizdagi kichik datasetga tayanib ishlasangiz (Real MLda 99% shu holat bo'ladi).
# Misol: ML model orqali yuborilgan push-bildirishnomalarni ochgan foydalanuvchilar o'rtacha vaqti (Guruh 1) va eski usulda yuborilganlar vaqti (Guruh 2).

#  B. ANOVA Test (Uch yoki undan ko'p guruhlar jangi)
# Qachon ishlatiladi: T-test faqat 2 ta guruhni solishtira olardi. Agar guruhlar soni 3 ta yoki undan ko'p bo'lsa, ANOVA ishga tushadi.
# Misol: Sayt foydalanuvchilarini 3 ta guruhga bo'ldingiz: Yoshlar, O'rta yoshlilar va Kattalar. Ularning saytda sarflagan o'rtacha pullarini (raqamli qiymat) solishtirmoqchisiz.

#  C. Chi-Square (\chi^2) Test (Matnli/Kategoriyali ustunlar bog'liqligi)
# Qachon ishlatiladi: Tepada hamma testlarda raqamlar (pullar, vaqtlar) bor edi. Agar qo'lingizda umuman raqam bo'lmasa-chi? Faqat matnli toifalar bo'lsa?
# Misol: Ustun 1: Jins (Erkak / Ayol). Ustun 2: Obuna turi (Premium / Bepul).
# Savol: Jins bilan obuna turi o'rtasida bog'liqlik bormi? Premium obunani erkaklar ko'proq olyaptimi yoki ayollarmi?
# Mana shu matnli toifalarni bir-biriga solishtirish uchun Chi-Square testi ishlatiladi.

import numpy as np
import scipy.stats as stats

# 1. Sun'iy ma'lumotlar yaratamiz (Mijozlarning xarid summasi, ming so'mda)
eski_tizim_A = np.array([120, 150, 110, 130, 140, 125, 135, 115, 145, 120])
yangi_tizim_B = np.array([140, 165, 150, 135, 170, 155, 160, 145, 150, 165])

print("==========================================================")
# H0 (Nol gipoteza): Yangi tizim sotuvga ta'sir qilmadi. O'rtacha qiymatlar teng.
# H1 (Muqobil gipoteza): Yangi tizim sotuvni oshirdi. O'rtachalar o'rtasida farq bor.

# 2. Mustaqil ikki tanlanma uchun T-test o'tkazamiz (Two-sample T-test)
t_statistika, p_value = stats.ttest_ind(eski_tizim_A, yangi_tizim_B)

print(f"T-statistika: {t_statistika:.4f}")
print(f"P-value (Tasodif ehtimoli): {p_value:.5f}")

# 3. Alfa = 0.05 chegarasiga ko'ra hukm chiqaramiz
alfa = 0.05
if p_value < alfa:
    print("\n🔥 Xulosa: P-value < 0.05! Nol gipoteza (H0) RAD ETILADI.")
    print("Yangi ML tizimi xaridlar summasini ROSTDAN HAM oshirgan. Natija tasodif emas!")
else:
    print("\n Xulosa: P-value >= 0.05. Nol gipoteza RAD ETILMAYDI.")
    print("Ikki tizim o'rtasida farq sezilarsiz, o'rtadagi farq shunchaki tasodif bo'lishi mumkin.")

    import numpy as np
import scipy.stats as stats

# 1. Guruhlar ma'lumotlari (Foydalanuvchilarning o'yinda qolgan vaqti, daqiqada)
fb_reklama = np.array([15, 20, 18, 22, 17])
yt_reklama = np.array([25, 28, 32, 24, 30])
tt_reklama = np.array([12, 14, 19, 11, 15])

print("==========================================================")
# H0: Uchala platformadan kelgan foydalanuvchilar o'rtacha vaqti bir xil.
# H1: Kamida bitta platformaning natijasi qolganlaridan keskin farq qiladi.

# 2. Bir faktorli ANOVA (One-way ANOVA) testini ishga tushiramiz
f_statistika, p_value = stats.f_oneway(fb_reklama, yt_reklama, tt_reklama)

print(f"F-statistika: {f_statistika:.4f}")
print(f"P-value (Tasodif ehtimoli): {p_value:.6f}")

# 3. Qaror qabul qilamiz
if p_value < 0.05:
    print("\n🔥 Xulosa: P-value < 0.05! Nol gipoteza (H0) RAD ETILADI.")
    print("Platformalar o'rtasida KESKIN farq bor! Reklamalar bir xil natija bermayapti.")
else:
    print("\n Xulosa: P-value >= 0.05. H0 RAD ETILMAYDI.")
    print("Uchala platformada ham deyarli bir xil natija. Farqlar shunchaki tasodifiy.")

import numpy as np
import pandas as pd
import scipy.stats as stats

# 1. Kesishma jadvalini (Contingency Table) tuzamiz
# Ustunlar: Telefon, Noutbuk | Satrlar: Erkak, Ayol
# Masalan: 120 ta erkak telefon, 50 ta erkak noutbuk sotib olgan
jadval = np.array([[120, 50], [60, 80]])

# Chiroyli ko'rinish uchun Pandas DataFrame ga o'tkazamiz
df = pd.DataFrame(
    jadval,
    index=["Erkak", "Ayol"],
    columns=["Telefon sotib olganlar", "Noutbuk sotib olganlar"],
)
print("=== TAHLIL QILINAYOTGAN MA'LUMOTLAR JADVALI ===")
print(df)
print("==========================================================")

# H0: Jins va mahsulot tanlovi o'rtasida HECH QANDAY bog'liqlik yo'q (Ular mustaqil).
# H1: Jins va mahsulot tanlovi o'rtasida KUCHLI bog'liqlik bor!

# 2. Chi-Square mustaqillik testini o'tkazamiz
chi2_stat, p_value, dof, expected = stats.chi2_contingency(jadval)

print(f"Chi-Square statistika: {chi2_stat:.4f}")
print(f"P-value (Tasodif ehtimoli): {p_value:.6f}")

# 3. Qaror qabul qilamiz
if p_value < 0.05:
    print("\n🔥 Xulosa: P-value < 0.05! Nol gipoteza (H0) RAD ETILADI.")
    print("Xaridorning jinsi va u tanlayotgan mahsulot o'rtasida QAT'IY bog'liqlik bor.")
    print("Reklama modelida jins omilini albatta hisobga olish kerak!")
else:
    print("\n Xulosa: P-value >= 0.05. H0 RAD ETILMAYDI.")
    print("Jins va mahsulot turi bir-biriga bog'liq emas. Erkaklar ham, ayollar ham deyarli bir xil ulushda sotib olishmoqda.")
```
